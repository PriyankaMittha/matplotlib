# Box plot

```python
import matplotlib.pyplot as plt
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x)
plt.show()
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, notch = True)
plt.show()
```

Horizontal

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, vert = False)
plt.show()
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, vert = True)
plt.show()
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, widths = 0.5)
plt.show()
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"])
plt.show()
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], patch_artist = True)
plt.show()
```

```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True)
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True)
plt.show()
```

```python
x = [10,15,3,40,25,60,70, 130,150]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True)
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True)
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True, sym = "r+")
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True, sym = "r+", boxprops = dict(color = "r"))
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 130, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True, sym = "r+", boxprops = dict(color = "r"), capprops = dict(color = "aqua"))
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 130, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True, sym = "r+", boxprops = dict(color = "r"), capprops = dict(color = "aqua"), whiskerprops = dict(color = "magenta"))
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 130, 140]
plt.boxplot(x, widths = 0.2, tick_labels = ["code"], showmeans = True, 
            boxprops = dict(color = "r"), capprops = dict(color = "aqua"), 
            whiskerprops = dict(color = "magenta"), flierprops = dict(marker='o',
            markeredgecolor='yellow', markerfacecolor='yellow'))
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 130, 140]
x1 = [10,20,30,40,50,60,70, 90, 120]

y = [x,x1]

plt.boxplot(y, widths = 0.2, tick_labels = ["apple", "cherry"], showmeans = True, 
            boxprops = dict(color = "r"), capprops = dict(color = "aqua"), 
            whiskerprops = dict(color = "magenta"), flierprops = dict(marker='o',
            markeredgecolor='yellow', markerfacecolor='yellow'))
plt.show()
```

```python
x = [10,20,30,40,50,60,70, 130, 140]
x1 = [10,20,30,40,50,60,70, 90, 120]
x2 = [25,23,28,30,48,50,60,75]

y = [x,x1,x2]

plt.boxplot(y, widths = 0.2, tick_labels = ["apple", "cherry", "orange"], showmeans = True, 
            boxprops = dict(color = "r"), capprops = dict(color = "aqua"), 
            whiskerprops = dict(color = "magenta"), flierprops = dict(marker='o',
            markeredgecolor='yellow', markerfacecolor='yellow'))
plt.show()
```

# Stack plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
area = [1,2,3,4,5]
plt.stackplot(x,area)

plt.show()
```

```python
x = [1,2,3,4,5]
area = [2,5,3,4,1]
plt.stackplot(x,area)

plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

plt.stackplot(x,area1,area2,area3)

plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l)
plt.legend()
plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l)
plt.legend(loc = 1)
plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l, colors = ['r', 'aqua', 'y'])
plt.legend(loc = 1)
plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l, colors = ['r', 'aqua', 'y'], baseline = "sym")
plt.legend(loc = 1)
plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l, colors = ['r', 'aqua', 'y'], baseline = "zero")
plt.legend(loc = 1)
plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l, colors = ['r', 'aqua', 'y'])
plt.title("Area Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend(loc = 1)
plt.show()
```

```python
x = [1,2,3,4,5]
area1 = [2,5,3,4,1]
area2 = [1,2,3,4,2]
area3 = [4,3,2,1,5]

l = ["area1", "area2", "area3"]
plt.stackplot(x,area1,area2,area3, labels = l, colors = ['r', 'aqua', 'y'])
plt.title("Area Plot", fontsize = 20)
plt.xlabel("x axis", fontsize = 20)
plt.ylabel("y axis", fontsize = 20)
plt.grid()
plt.legend(loc = 1)
plt.show()
```
