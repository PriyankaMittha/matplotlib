# Step plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [1,2,3,4,5]

plt.step(x,y)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y)
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y)
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y, color = 'r')
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y, color = 'r', marker = 'o')
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y, color = 'r', marker = 'o', ms = 15)
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y, color = 'r', marker = 'o', ms = 15, mfc = 'blue')
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y, color = 'r', marker = 'o', ms = 15, mfc = 'blue', label = "price")
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)

plt.legend()
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,3,5,1]

plt.step(x,y, color = 'r', marker = 'o', ms = 15, mfc = 'blue', label = "price")
plt.title("Step plot", fontsize = 15)
plt.xlabel("X axis", fontsize = 15)
plt.ylabel("Y axis", fontsize = 15)

plt.legend(loc = 2)
plt.grid()
plt.show()
```

# Fill Between plot

```python
x = [1,2,3,4,5]
y = [1,2,3,4,5]

plt.plot(x,y, color = 'y')
plt.fill_between(x,y)
plt.title("Fill Between Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.show()
```

```python
x = [1,2,3,4,5]
y = [1,2,3,4,5]

plt.plot(x,y, color = 'y')
plt.fill_between(x=[2,4],y1=2,y2=4)
plt.title("Fill Between Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.show()
```

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([1,2,3,4,5])

plt.plot(x,y, color = 'y')
plt.fill_between(x,y, where = (x>=2) & (x<=4))
plt.title("Fill Between Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.show()
```

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([1,2,3,4,5])

plt.plot(x,y, color = 'y')
plt.fill_between(x,y, where = (x>=3) & (x<=4.5))
plt.title("Fill Between Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.show()
```

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([1,2,3,4,5])

plt.plot(x,y, color = 'y')
plt.fill_between(x,y, where = (x>=2) & (x<=4), alpha = 0.4)
plt.title("Fill Between Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.show()
```

# Subplot 

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [1,2,3,4]
plt.plot(x,y,color = 'r')

x1 = [1,2,3,4,5]
y1 = [2,3,5,1,4]
plt.step(x,y)

x3 = [10,20,30,40]
plt.pie(x3)

x4 = ['a', 'b', 'c', 'd']
y4 = [2,3,4,5]
plt.bar(x4,y4)
plt.show()
```

```python
x = [1,2,3,4]
y = [1,2,3,4]
plt.plot(x,y,color = 'r')
plt.subplot(2,2,1)
x1 = [1,2,3,4,5]
y1 = [2,3,5,1,4]
plt.step(x,y)
plt.subplot(2,2,2)
x3 = [10,20,30,40]
plt.pie(x3)
plt.subplot(2,2,3)
x4 = ['a', 'b', 'c', 'd']
y4 = [2,3,4,5]
plt.bar(x4,y4)
plt.subplot(2,2,4)

plt.show()
```

# Text

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,5,1,3]
plt.plot(x,y,color = 'r')
plt.title("Growth Graph")
plt.text(2,4,"apple", fontsize = 15, style = "italic")

plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,5,1,3]
plt.plot(x,y,color = 'r')
plt.title("Growth Graph")
plt.text(2,3,"apple", fontsize = 15, style = "italic", bbox = {"facecolor": 'y'})

plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,5,1,3]
plt.plot(x,y,color = 'r')
plt.title("Growth Graph")
plt.text(2,4,"apple", fontsize = 15, style = "italic")

plt.annotate("student", xy=(2,1), xytext=(4,4), arrowprops = dict(facecolor = 'g', shrink = 100))

plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,5,1,3]
plt.plot(x,y,color = 'r')
plt.title("Growth Graph")
plt.text(2,4,"apple", fontsize = 15, style = "italic")

plt.annotate("student", xy=(3,4), xytext=(4,4), arrowprops = dict(facecolor = 'g', shrink = 100))

plt.show()
```
